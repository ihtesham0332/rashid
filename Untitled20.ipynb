```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

# 1. Load dataset
df = pd.read_csv("updated local data.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# 2. Create target: Low, Medium, High
df["severity"] = pd.cut(
    df["Hba1c"],
    bins=[-np.inf, 5.7, 6.5, np.inf],
    labels=["Low", "Medium", "High"],
    right=False
)

# 3. Remove rows with missing target/features
df = df.dropna(
    subset=[
        "Age", "gender", "Glucose", "weight", "height",
        "bmi", "systolic_bp", "diastolic_bp",
        "Hba1c", "severity"
    ]
)

# 4. Features: HbA1c INCLUDED
features = [
    "Age", "gender", "Glucose", "weight", "height",
    "bmi", "systolic_bp", "diastolic_bp", "Hba1c"
]

X = df[features]
y = df["severity"]

# 5. Preprocessing
numeric_features = [
    "Age", "Glucose", "weight", "height", "bmi",
    "systolic_bp", "diastolic_bp", "Hba1c"
]
categorical_features = ["gender"]

preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numeric_features),
    ("categorical", OneHotEncoder(handle_unknown="ignore"),
     categorical_features)
])

# 6. Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# 7. SVM model
svm_model = Pipeline([
    ("preprocessor", preprocessor),
    ("svm", SVC(kernel="rbf", C=1.0, gamma="scale"))
])

svm_model.fit(X_train, y_train)

# 8. Predictions
y_pred = svm_model.predict(X_test)

print("\nSVM RESULTS WITH HbA1c")
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Macro F1:", round(f1_score(
    y_test, y_pred, average="macro"
), 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test, y_pred,
    labels=["Low", "Medium", "High"]
))

# 9. 5-fold cross-validation
cv_scores = cross_val_score(
    svm_model, X, y,
    cv=5,
    scoring="f1_macro"
)

print("\n5-Fold CV Macro F1:", cv_scores)
print("Mean CV Macro F1:", round(cv_scores.mean(), 4))
print("CV Standard Deviation:", round(cv_scores.std(), 4))
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

# 1. Load dataset
df = pd.read_csv("updated local data.csv")

# 2. Create severity target using HbA1c
df["severity"] = pd.cut(
    df["Hba1c"],
    bins=[-np.inf, 5.7, 6.5, np.inf],
    labels=["Low", "Medium", "High"],
    right=False
)

df = df.dropna(subset=[
    "Age", "gender", "Glucose", "weight", "height",
    "bmi", "systolic_bp", "diastolic_bp", "Hba1c", "severity"
])

# 3. Features WITHOUT HbA1c
features = [
    "Age", "gender", "Glucose", "weight", "height",
    "bmi", "systolic_bp", "diastolic_bp"
]

X = df[features]
y = df["severity"]

numeric_features = [
    "Age", "Glucose", "weight", "height", "bmi",
    "systolic_bp", "diastolic_bp"
]
categorical_features = ["gender"]

preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numeric_features),
    ("categorical", OneHotEncoder(handle_unknown="ignore"),
     categorical_features)
])

# 4. Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 5. SVM pipeline
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("svm", SVC())
])

# 6. Hyperparameter tuning
param_grid = {
    "svm__C": [0.1, 1, 10, 100],
    "svm__kernel": ["rbf", "linear"],
    "svm__gamma": ["scale", "auto"],
    "svm__class_weight": [None, "balanced"]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

grid = GridSearchCV(
    pipeline,
    param_grid,
    cv=cv,
    scoring="f1_macro",
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)

# 7. Evaluate on test set
best_svm_no_hba1c = grid.best_estimator_
y_pred = best_svm_no_hba1c.predict(X_test)

print("\n===== SVM WITHOUT HbA1c =====")
print("Features:", features)
print("Best Parameters:", grid.best_params_)
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Macro F1:", round(
    f1_score(y_test, y_pred, average="macro"), 4
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("Confusion Matrix (Low, Medium, High):")
print(confusion_matrix(
    y_test, y_pred,
    labels=["Low", "Medium", "High"]
))

print("\nBest 5-Fold CV Macro F1:",
      round(grid.best_score_, 4))
```
